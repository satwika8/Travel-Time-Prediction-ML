```python
"""
Simple Random Forest Example
============================
A small, easy-to-follow example: load a toy dataset, train a
Random Forest, check accuracy, OOB score, and feature importances.
"""

import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report

RANDOM_STATE = 42

# ---------------------------------------------------------------
# 1. Load the simple dataset
# ---------------------------------------------------------------
df = pd.read_csv("train.csv")

print("Sample of the dataset:")
print(df.head())
print(f"\nDataset shape: {df.shape}")

TARGET = "Survived"
print(f"Class balance:\n{df[TARGET].value_counts()}")

DROP_COLS = ["PassengerId", "Name", "Ticket", "Cabin", "Embarked"]  # PassengerId, Name, Ticket, Cabin are just identifiers, Embarked needs preprocessing or dropping for simplicity

feature_names = [c for c in df.columns if c not in [TARGET] + DROP_COLS]

# Handle categorical features if any, for now drop them. For simplicity, we will drop non-numeric columns.
# In a real scenario, these would need proper encoding (e.g., one-hot encoding).
feature_names = [col for col in feature_names if df[col].dtype in ['int64', 'float64']]
X = df[feature_names].values
y = df[TARGET].values

# ---------------------------------------------------------------
# 2. Train / test split
# ---------------------------------------------------------------
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=RANDOM_STATE
)

# ---------------------------------------------------------------
# 3. Train a Random Forest
# ---------------------------------------------------------------
rf = RandomForestClassifier(
    n_estimators=100,
    max_features="sqrt",
    oob_score=True,
    random_state=RANDOM_STATE,
)
rf.fit(X_train, y_train)

# ---------------------------------------------------------------
# 4. Evaluate
# ---------------------------------------------------------------
y_pred = rf.predict(X_test)
test_accuracy = accuracy_score(y_test, y_pred)

print(f"\nOOB score: {rf.oob_score_:.4f}")
print(f"Test accuracy: {test_accuracy:.4f}")
print("\nClassification report:")
print(classification_report(y_test, y_pred))

# ---------------------------------------------------------------
# 5. Feature importances
# ---------------------------------------------------------------
print("Feature importances:")
for name, importance in sorted(
    zip(feature_names, rf.feature_importances_), key=lambda x: -x[1]
):
    print(f"  {name}: {importance:.4f}")
```
