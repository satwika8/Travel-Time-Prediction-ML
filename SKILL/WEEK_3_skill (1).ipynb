```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.backends.backend_pdf import PdfPages

sns.set_style("whitegrid")
```

```python
df = pd.read_csv("placement_predict_50k.csv")
```

```python
num_cols = ['CGPA' , 'AttendancePercent' , 'AptitudeTestScore' , 'CodingTestScore' , 'SoftSkillsRating' , 'SalaryPackage']
```

```python
import seaborn as sns
fig1 = plt.figure(figsize=(10, 6))
sns.heatmap(df.isnull(), cbar=False, cmap="Purples", yticklabels=False)
plt.title("Missingness Heatmap - Placement Prediction Dataset")
plt.xlabel("Columns")
plt.show()
```

```python
fig2, axes = plt.subplots(2, 3, figsize=(15, 8))

print("Columns in DataFrame df:", df.columns.tolist())

for i, col in enumerate(num_cols):
    if col not in df.columns:
        print(f"Warning: Column '{col}' not found in DataFrame. Skipping.")
        continue

    row, colpos = divmod(i, 3)
    sns.histplot(df[col].dropna(), bins=30, kde=True,
                 ax=axes[row, colpos], color='steelblue')
    axes[row, colpos].set_title(f'Distribution of {col}')

plt.tight_layout()
plt.show()
```
